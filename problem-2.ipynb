## Part C

1. Exact solution of toy-model problem with C_l = 3.40665e-3

```python
import numpy as np
from scipy.optimize import fsolve
import matplotlib.pyplot as plt

a_v = 15.71511
a_s = 17.53638
a_c = 0.71363
a_a = 23.37837
delta_m = 1.3
C_l = 3.40665e-3

def eq_system(p, p_F_val):
    x, y = p
    
    # (Equation 27)
    a_c_tilde = a_c - C_l * p_F_val
    
    # To prevent errors
    if x <= 0 or y <= 0:
        return [1e6, 1e6]
        
    eq1 = 2 * a_c_tilde * x * y**2 - a_s / x**2
    eq2 = -delta_m + 2 * a_c_tilde * (x**2) * y - 4 * a_a * (1 - 2*y) + (y**(1/3)) * p_F_val
    
    return [eq1, eq2]

p_F_range = np.linspace(0, 40, 400)

x_solutions = []
y_solutions = []

# Initial guess for the very first iteration
current_guess = [4.0, 0.4]

# Loop through each p_F value
for p_F_val in p_F_range:
    # Solve using the last successful solution as the next initial guess
    sol, info, ier, msg = fsolve(eq_system, current_guess, args=(p_F_val,), full_output=True)
    
    if ier == 1:  # Solution converged successfully
        x_solutions.append(sol[0])
        y_solutions.append(sol[1])
        current_guess = [sol[0], sol[1]]  # Update guess for the next step
    else:
        # If it fails to converge, break the loop or append NaN to handle it gracefully
        x_solutions.append(np.nan)
        y_solutions.append(np.nan)

x_solutions_1 = np.array(x_solutions)
y_solutions_1 = np.array(y_solutions)

```

2. Exact solution of toy-model problem with C_l = 0

```python
import numpy as np
from scipy.optimize import fsolve
import matplotlib.pyplot as plt

a_v = 15.71511
a_s = 17.53638
a_c = 0.71363
a_a = 23.37837
delta_m = 1.3
C_l = 0

def eq_system(p, p_F_val):
    x, y = p
    
    # (Equation 27)
    a_c_tilde = a_c - C_l * p_F_val
    
    # To prevent errors
    if x <= 0 or y <= 0:
        return [1e6, 1e6]
        
    eq1 = 2 * a_c_tilde * x * y**2 - a_s / x**2
    eq2 = -delta_m + 2 * a_c_tilde * (x**2) * y - 4 * a_a * (1 - 2*y) + (y**(1/3)) * p_F_val
    
    return [eq1, eq2]

p_F_range = np.linspace(0, 40, 400)

x_solutions = []
y_solutions = []

# Initial guess for the very first iteration
current_guess = [4.0, 0.4]

# Loop through each p_F value
for p_F_val in p_F_range:
    # Solve using the last successful solution as the next initial guess
    sol, info, ier, msg = fsolve(eq_system, current_guess, args=(p_F_val,), full_output=True)
    
    # If the solver converges (ier == 1), store the results and update the next guess
    if ier == 1:  # Solution converged successfully
        x_solutions.append(sol[0])
        y_solutions.append(sol[1])
        current_guess = [sol[0], sol[1]]  # Update guess for the next step
    else:
        # If it fails to converge, break the loop or append NaN to maintain list length
        x_solutions.append(np.nan)
        y_solutions.append(np.nan)

x_solutions_2 = np.array(x_solutions)
y_solutions_2 = np.array(y_solutions)


```
