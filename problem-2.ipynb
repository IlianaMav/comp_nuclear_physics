## Part A

From eq.3 to eq.4

```python
p = sp.Symbol('p')
pfe=sp.Symbol('pfe')
me=sp.Symbol('me')
n=sp.Symbol('n')
pi = sp.pi
xf=sp.Symbol('xf')
yf=sp.Symbol('yf')
```

```python
constant = 1/(n * pi**2)

f = (p**2) * (sp.sqrt(p**2+me**2))

integral = sp.integrate(f, (p,0,pfe))

Ee = constant * integral
```

```python
print('Integral=')
print(integral)
print('---------------------------------------------------------------------------------------------------------')
print('Ee=')
print(Ee)
```

```python
Ee = Ee.rewrite(sp.log)
print('Ee=')
print(Ee)
```

```python
Ee= Ee.subs(pfe,me*xf)
print(Ee)
```

```python
Ee =Ee.subs(sp.sqrt(1+xf**2),yf)
print(Ee)
```

```python
Ee = sp.factor(Ee)
print(Ee)
```

```python
sp.pprint(Ee)
```

## Part B

1. From eq.19 to eq.21

```python
mp = sp.Symbol('mp')
mn = sp.Symbol('mn')

a_v = sp.Symbol('a_v')
a_c = sp.Symbol('a_c')
a_s = sp.Symbol('a_s')
a_a = sp.Symbol('a_a')

x = sp.Symbol('x')
y = sp.Symbol('y')
```

```python
f = mp*y + mn*(1-y) - a_v + (a_s/x) + a_c*x**2*y**2 + a_a*(1-2*y)**2
print(f)
```

```python
thetaf_x = sp.diff(f,x)
print('thetaf_x=')
print(thetaf_x)
```

```python
thetaf_y = sp.diff(f,y)
print('thetaf_y=')
print(thetaf_y)
```

2. Solve System of equations

```python
solutions = sp.solve((thetaf_x,thetaf_y), (x,y))

for _,solution in enumerate(solutions):
    print(solution)
```

```python
solutions = sp.nonlinsolve((thetaf_x,thetaf_y), (x,y))

for _,solution in enumerate(solutions):
    print(solution)
```

```python
params = {a_v:15.71511, a_s:17.53638, a_c:0.71363, a_a:23.37837}

solutions = sp.nsolve((thetaf_x.subs(params),thetaf_y.subs(params)), (x,y), (1,1))

for _,solution in enumerate(solutions):
    print(solution)
```
