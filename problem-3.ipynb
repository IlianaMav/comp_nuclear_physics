```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
```

## (1) lamda_1 < lamda_2

```python
N0 = 1000
lamda_1 = 0.05
lamda_2 = 0.3

t = np.linspace(0, 100, 500)
```

```python
N1 = N0 * np.exp(-lamda_1 * t)

N2 = (lamda_1 / (lamda_2 - lamda_1)) * N0 * \
         (np.exp(-lamda_1 * t) - np.exp(-lamda_2 * t))

term1 = (1 - np.exp(-lamda_1 * t)) / lamda_1
term2 = (1 - np.exp(-lamda_2 * t)) / lamda_2

N3 = ((lamda_1 * lamda_2) / (lamda_2 - lamda_1)) * N0 * (term1 - term2)

plt.figure(figsize=(9, 5))

plt.plot(t, N1, label='N1', linewidth=2, color='blue')
plt.plot(t, N2, label='N2', linewidth=2, color='red')
plt.plot(t, N3, label='N3', linewidth=2, color='green')

plt.xlabel('Time')
plt.ylabel('Number of Nuclei')
plt.title(f'Analytical Solution with lamda_1={lamda_1}, lamda_2={lamda_2}')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```

```python
from scipy.integrate import odeint

def system(N, t, lamda1, lamda2):

    N1,N2,N3=N

    dN1_dt = -lamda1 * N1
    dN2_dt = lamda1 * N1 - lamda2 * N2
    dN3_dt = lamda2 * N2

    return [dN1_dt, dN2_dt, dN3_dt]

initial_conditions = [N0, 0, 0]

solution = odeint(system, initial_conditions, t, args=(lamda_1, lamda_2))

N1_num = solution[:, 0]
N2_num = solution[:, 1]
N3_num = solution[:, 2]



plt.figure(figsize=(9, 5))

plt.plot(t, N1_num, label='N1', linewidth=2, color='blue')
plt.plot(t, N2_num, label='N2', linewidth=2, color='red')
plt.plot(t, N3_num, label='N3', linewidth=2, color='green')

plt.xlabel('Time')
plt.ylabel('Number of Nuclei')
plt.title(f'Numerical Solution with lamda_1={lamda_1}, lamda_2={lamda_2}')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint


def chain_5(N, t, l1, l2, l3, l4):
    N1, N2, N3, N4, N5 = N
    
    dN1_dt = -l1 * N1
    dN2_dt = l1 * N1 - l2 * N2
    dN3_dt = l2 * N2 - l3 * N3
    dN4_dt = l3 * N3 - l4 * N4
    dN5_dt = l4 * N4 # Ο N5 είναι ο σταθερός τελικός πυρήνας
    
    return [dN1_dt, dN2_dt, dN3_dt, dN4_dt, dN5_dt]

N0 = 1000
l1, l2, l3, l4 = 0.5, 0.3, 0.1, 0.05
t = np.linspace(0, 100, 500)
initial = [N0, 0, 0, 0, 0]

sol = odeint(chain_5, initial, t, args=(l1, l2, l3, l4))

plt.figure(figsize=(10, 6))
labels = ['N1', 'N2', 'N3', 'N4', 'N5']
colors = ['blue', 'red', 'orange', 'purple', 'green']

for i in range(5):
    plt.plot(t, sol[:, i], label=labels[i], linewidth=2, color=colors[i])

plt.xlabel('Time')
plt.ylabel('Number of Nuclei')
plt.title(f'Numerical Solution')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
